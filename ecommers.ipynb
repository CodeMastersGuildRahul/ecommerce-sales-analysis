```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.replace(['N/A','NULL',''],np.nan,inplace=True)
df
```

```python
#remove spaces
df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x) 
```

```python
#standardize Text format
#convert text into proper case

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()

df
```

```python
#validation email address

df = df[df['Email'].str.contains('@',na=False)]
```

```python
#Convert date column

df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)

df['Delivery_Date'] = pd.to_datetime(
    df['Delivery_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)
```

```python
#convert numaric columns

cols = ['Qty','Unit_Price','Discount']

for c in cols : 
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
#remove invalide quentity
#remove record having negative or zero quantity
df = df[df['Qty']>0]
```

```python
#full missing value

# Create an independent DataFrame
df = df.copy()

# Handle missing values
df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].astype('string').fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
#Feature Engineering
#Total sale

df['Sales']=df['Qty']*df['Unit_Price']
df
```

```python
#Net Amount

df['Net_Amount'] = df['Sales']-(df['Sales']*df['Discount']/100)
df
```

```python
#profit(Example: 20%)
df['Profit']=df['Net_Amount']*0.20
df
```

```python
#Order Month

df['Month'] = df['Order_Date'].dt.month_name()
df
```

```python
#order year
df['Year'] = df['Order_Date'].dt.year
df
```

```python
df['Weekday'] = df['Order_Date'].dt.day_name()
df
```

```python
#Final Data visualation
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
#Export clean Data
df.to_csv('Clean_Ecommerce.csv',index=False)
```
